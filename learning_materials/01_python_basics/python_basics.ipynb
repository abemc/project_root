## 1. 変数とデータ型

```python
# 基本的な型
x = 42
y = 3.14
name = "Alice"
flag = True
nothing = None

for var_name, val in [("x", x), ("y", y), ("name", name), ("flag", flag), ("nothing", nothing)]:
    print(f"{var_name:10s} = {val!r:10} (型: {type(val).__name__})")
```

```python
# コレクション型
fruits = ["apple", "banana", "cherry"]
fruits.append("date")
print("リスト:", fruits)
print("スライス:", fruits[1:3])

coords = (35.6895, 139.6917)  # 緯度・経度（変更不可）
print("タプル:", coords)

person = {"name": "Alice", "age": 30}
person["email"] = "alice@example.com"
print("辞書:", person)
print("安全なアクセス:", person.get("phone", "N/A"))

numbers = {1, 2, 3, 2, 1}  # 重複は除去される
print("集合:", numbers)
```

## 2. 制御構造

```python
# 条件分岐
scores = [45, 62, 78, 85, 95]
for score in scores:
    grade = "A" if score >= 90 else "B" if score >= 75 else "C" if score >= 60 else "F"
    print(f"  {score}点 → {grade}")
```

```python
# ループとリスト内包表記
squares = [x**2 for x in range(1, 11)]
print("二乗:", squares)

evens = [x for x in range(20) if x % 2 == 0]
print("偶数:", evens)

# enumerate で添字付き
fruits = ["りんご", "バナナ", "さくらんぼ"]
for i, fruit in enumerate(fruits, start=1):
    print(f"  {i}. {fruit}")
```

## 3. 関数とデコレータ

```python
def greet(name: str, greeting: str = "こんにちは") -> str:
    """挨拶文を返す関数。"""
    return f"{greeting}、{name}さん！"

print(greet("Alice"))
print(greet("Bob", "おはよう"))

# *args と **kwargs
def summarize(*args, **kwargs):
    print("位置引数:", args)
    print("キーワード引数:", kwargs)

summarize(1, 2, 3, name="Alice", age=30)
```

```python
import time
import functools

def timer(func):
    """実行時間を計測するデコレータ。"""
    @functools.wraps(func)
    def wrapper(*args, **kwargs):
        start = time.perf_counter()
        result = func(*args, **kwargs)
        elapsed = time.perf_counter() - start
        print(f"  ⏱ {func.__name__} の実行時間: {elapsed*1000:.2f}ms")
        return result
    return wrapper

@timer
def sum_range(n: int) -> int:
    return sum(range(n))

result = sum_range(1_000_000)
print(f"  sum(0..999999) = {result:,}")
```

## 4. ファイル操作（CSV / JSON）

```python
import csv
from pathlib import Path

# CSV を書き込んで読み込む
csv_path = Path("sample_people.csv")

rows = [
    {"name": "Alice", "age": 30, "city": "Tokyo"},
    {"name": "Bob",   "age": 25, "city": "Osaka"},
    {"name": "Carol", "age": 35, "city": "Kyoto"},
]

with csv_path.open("w", newline="", encoding="utf-8") as f:
    writer = csv.DictWriter(f, fieldnames=["name", "age", "city"])
    writer.writeheader()
    writer.writerows(rows)

print("CSV 読み込み結果:")
with csv_path.open(encoding="utf-8") as f:
    for row in csv.DictReader(f):
        print(f"  {row['name']:8s} | {row['age']:2s}歳 | {row['city']}")

csv_path.unlink()  # 後始末
```

```python
import json

data = {
    "project": "AI 学習ロードマップ",
    "version": "1.0",
    "topics": ["Python", "数学", "機械学習", "深層学習"],
    "completed": False,
}

json_str = json.dumps(data, ensure_ascii=False, indent=2)
print(json_str)

# 文字列から読み戻す
loaded = json.loads(json_str)
print("\nプロジェクト名:", loaded["project"])
print("トピック数:", len(loaded["topics"]))
```
